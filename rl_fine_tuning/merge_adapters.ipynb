```python
from huggingface_hub import login
import os 
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import PeftModel

with open('/project/hf_api_login_key.txt', 'r') as f:
    hf_api = f.read()
login(token=hf_api)


# === CONFIGURATION ===
# base model from Hugging Face
BASE_MODEL = "imge/rf_llama_merged_llms_v1"

# path to LoRA adapter on HF
ADAPTER_PATH = "imge/reinforce_llama_adapters_llms_v2"

# === MERGE PROCESS ===

# Step 1: Load base model
print("🔹 Loading base model...")
base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    device_map="auto",
    trust_remote_code=True
)

# Step 2: Load and apply adapter
print("🔹 Loading adapter...")
peft_model = PeftModel.from_pretrained(
    base_model,
    ADAPTER_PATH,
    device_map="auto"
)

# Step 3: Merge LoRA weights into base model
print("🔹 Merging adapter into base model...")
merged_model = peft_model.merge_and_unload()

print("✅ Merge complete. Your merged model is ready!")

print("🔹 Pushing merged model to Hub...")
merged_model.push_to_hub("imge/rf_llama_merged_llms_v2")

print("🔹 Pushing tokenizer to Hub...")
tokenizer.push_to_hub("imge/rf_llama_merged_llms_v2")

print("✅ All done!")
```
